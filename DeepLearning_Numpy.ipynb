```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

true_b = 1
true_w = 2
N = 100 # 100 Data points

# Data generation
np.random.seed(42)
X = np.random.rand(N, 1)
epsilon = (.1 * np.random.randn(N, 1)) # introduce 1% noise
y = true_w * X + true_b + epsilon
# Split the data
train_split = int(0.8 * len(X))
x_train, y_train = X[:train_split], y[:train_split] # Train on 80% of the data
x_test, y_test = X[train_split:], y[train_split:]   # Test on the remaining 20% of the data

# Step 0 - Initialize parameters 'w' and 'b'
np.random.seed(42)
b = np.random.randn(1)
w = np.random.randn(1)

print(f"Original b & w: {b}, {w}")
# Step 1 Compute model's prediction, known as forward pass
yhat = w * x_train + b
# Compute the error
error = yhat - y_train
# Step 2 Compute the loss, which is the Mean Squared Error
loss = (error ** 2).mean()
print(f"Loss: {loss}")
# Step 3 Compute the gradient (or partial derivatives) for 'b & w'
b_grad = 2 * error.mean()
w_grad = 2 * (x_train * error).mean()
print(f"Gradients: {b_grad}, {w_grad}")
# Choose a learning rate
learning_rate = 0.1
# Step 4 Update the parameters 'b' and 'w'
b = b - learning_rate * b_grad
w = w - learning_rate * w_grad
print(f"Updated b & w: {b}, {w}")
# Step 5 Training loop (repeat above steps)
num_epochs = 1000
for epoch in range(num_epochs):
     # Compute predicted output
    yhat = b + w * x_train

    # Compute the error
    error = yhat - y_train
    loss = (error ** 2).mean()  # Compute MSE 
    # Compute the gradients for w & b
    b_grad = 2 * error.mean()
    w_grad = 2 * (x_train * error).mean()

    # Update parameters
    b = b - learning_rate * b_grad
    w = w - learning_rate * w_grad

print(f"Updated w & b after training loop, {w}, {b}")
    
```
